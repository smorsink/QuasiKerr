```python
import numpy as np
import subprocess as sub
import os
J_mev = 1.6021766e-13
cm_fm = 1e-13
erg_J = 1e7
c = 2.99792458e10
```

```python
def generate_eosfile(EOS):

    data = np.loadtxt("./code-compose-"+EOS+"/eos.table")

    P_mev_fm3 = data[:,3]
    rho_mev_fm3 = data[:,4]

    P_cgs = P_mev_fm3 * J_mev*erg_J/cm_fm**3
    rho_cgs = rho_mev_fm3 * J_mev*erg_J/c**2/cm_fm**3

    i_c = np.where(P_cgs > 0)[0]
    N = i_c.size

    if (P_cgs.size > N):
        print(str(P_cgs.size-N)+' lines removed from '+EOS)

    newtable = np.zeros([N,2])
    newtable[:,0] = rho_cgs[i_c]
    newtable[:,1] = P_cgs[i_c]
    tempfile = "eos_eP.table"
    np.savetxt(tempfile, newtable)

    command = "./hng.exe "+tempfile+" "+str(N)
    result = sub.run(command, stdout=sub.PIPE, text=True)

    eosfile = open("eos"+EOS, 'w')
    eosfile.write(result.stdout)
    eosfile.close()

    os.remove(tempfile)

    return N
```

```python
generate_eosfile('APR')
generate_eosfile('BBB')
generate_eosfile('BLC')
generate_eosfile('CMF1')
generate_eosfile('CMF1H')
generate_eosfile('CMF2')
generate_eosfile('D1MStar')
generate_eosfile('DD2Yd')
generate_eosfile('DDHdY4')
generate_eosfile('DDMEX')
generate_eosfile('GDFMII')
generate_eosfile('KDE0v')
generate_eosfile('M3')
generate_eosfile('NL3')
generate_eosfile('PCSB2')
generate_eosfile('PKDD')
generate_eosfile('QHC21A')
generate_eosfile('QMCRMF3')
generate_eosfile('Rs')
generate_eosfile('SK255')
generate_eosfile('SKa')
generate_eosfile('TW')
```
