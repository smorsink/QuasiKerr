```python
import numpy as np
import scipy.interpolate as terp
k = 1.3806503e-23
J_mev = 1.6021766e-13
cm_fm = 1e-13
erg_J = 1e7
c = 3e10
```

```python
data = np.loadtxt("eos.table")

nb_fm3 = data[:,1]
P_mev_fm3 = data[:,3]
rho_mev_fm3 = data[:,4]
```

```python
nb_cgs = nb_fm3 / cm_fm**3
P_cgs = P_mev_fm3 * J_mev*erg_J/cm_fm**3
rho_cgs = rho_mev_fm3 * J_mev*erg_J/c**2/cm_fm**3
```

```python
def integrate_h(rho, P):

    N = 5000
    h = np.zeros(rho.size)
    cs_rho = terp.CubicSpline(np.arange(rho.size), rho_cgs)
    cs_P = terp.CubicSpline(np.arange(rho.size), P_cgs)
    rho_val = cs_rho(np.linspace(0, rho.size, N))
    P_val = cs_P(np.linspace(0, rho.size, N))

    for i in range(rho.size):
        dP = P[i]/N
        integr = 0
        for k in range(N):
            dH = dP/(rho_val[k]+P_val[k])
            integr += dH
        h[i] = integr+1

    return h
```

```python
h_cgs = integrate_h(rho_cgs, P_cgs)
```
